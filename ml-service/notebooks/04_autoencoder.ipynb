# 04. Deep Autoencoder Anomaly Detection (PyTorch)
Trains a PyTorch Autoencoder to reconstruct normal transaction patterns and detect anomalies via MSE loss spikes.

```python
import torch
import torch.nn as nn

class Autoencoder(nn.Module):
    def __init__(self, input_dim):
        super(Autoencoder, self).__init__()
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 8),
            nn.ReLU(),
            nn.Linear(8, 4),
            nn.ReLU()
        )
        self.decoder = nn.Sequential(
            nn.Linear(4, 8),
            nn.ReLU(),
            nn.Linear(8, input_dim)
        )
    def forward(self, x):
        encoded = self.encoder(x)
        return self.decoder(encoded)

print('Autoencoder architecture defined successfully.')
```
