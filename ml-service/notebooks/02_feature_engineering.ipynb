# 02. Feature Engineering Pipeline
Generates origin and destination balance differentials, zero-balance flags, and one-hot encoded transaction types.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/paysim_subset.csv')

# Balance Deltas & Zero Flags
df['balanceDiffOrig'] = df['oldbalanceOrg'] - df['newbalanceOrig']
df['balanceDiffDest'] = df['newbalanceDest'] - df['oldbalanceDest']
df['origBalanceZero'] = (df['newbalanceOrig'] == 0).astype(int)
df['destBalanceZero'] = (df['oldbalanceDest'] == 0).astype(int)

# One-Hot Encoding
df = pd.get_dummies(df, columns=['type'], prefix='type')
df.head()
```
